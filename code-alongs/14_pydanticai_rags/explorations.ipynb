## pdfs in python

```python
from pypdf import PdfReader

reader = PdfReader("data/Dwarf_Hotot.pdf")
reader
```

```python

def extract_text_from_pdf(path):
    
    all_text = ""
    
    reader = PdfReader(path)

    for page in reader.pages:
        text = page.extract_text()

        all_text += text + "\n"

    return all_text

def export_text(text, export_path):
    with open(export_path, "w") as file:
        file.write(text)

dwarf_hotot = extract_text_from_pdf("data/Dwarf_Hotot.pdf")

export_text(dwarf_hotot, "data/dwarf_hotot.txt")
```
